```python
#checking sku distribution

sales.groupby("sku_id")["units_sold"].sum().sort_values(ascending=False).head(10)
```

```python

sku_sales = sales.groupby("sku_id")["units_sold"].sum().sort_values(ascending=False).head(10)

sku_sales_pct = sku_sales/sku_sales.sum()

sku_cumsum = sku_sales_pct.cumsum()

(sku_cumsum <= 0.8).sum()
```
